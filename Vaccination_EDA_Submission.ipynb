```python
dtp = (
    coverage[coverage["Antigen"].isin(["DTP1", "DTP3"])]
    .groupby(["Code", "Name", "Year", "Antigen"], as_index=False)["Coverage"]
    .mean()
)

dtp_pivot = dtp.pivot_table(
    index=["Code", "Name", "Year"],
    columns="Antigen",
    values="Coverage"
).reset_index()

dtp_pivot["DTP_Dropoff_Percentage_Points"] = (
    dtp_pivot["DTP1"] - dtp_pivot["DTP3"]
)

display(dtp_pivot.head())

print(
    "Average DTP1 to DTP3 drop-off:",
    round(dtp_pivot["DTP_Dropoff_Percentage_Points"].mean(), 2),
    "percentage points"
)
```

```python
dtp_year = (
    dtp_pivot.groupby("Year", as_index=False)["DTP_Dropoff_Percentage_Points"]
    .mean()
)

plt.figure(figsize=(10, 5))
plt.plot(
    dtp_year["Year"],
    dtp_year["DTP_Dropoff_Percentage_Points"],
    marker="o"
)

plt.title("Average DTP1 to DTP3 Drop-off by Year")
plt.xlabel("Year")
plt.ylabel("Drop-off (Percentage Points)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

display(dtp_year)
```

### Insight
The DTP1 to DTP3 drop-off measures how much vaccination coverage decreases between the first and third doses. A larger positive value indicates that fewer individuals complete the full DTP vaccination sequence.

```python
mcv = (
    coverage[coverage["Antigen"].isin(["MCV1", "MCV2"])]
    .groupby(["Code", "Name", "Year", "Antigen"], as_index=False)["Coverage"]
    .mean()
)

mcv_pivot = mcv.pivot_table(
    index=["Code", "Name", "Year"],
    columns="Antigen",
    values="Coverage"
).reset_index()

mcv_pivot["MCV_Dropoff_Percentage_Points"] = (
    mcv_pivot["MCV1"] - mcv_pivot["MCV2"]
)

display(mcv_pivot.head())

print(
    "Average MCV1 to MCV2 drop-off:",
    round(mcv_pivot["MCV_Dropoff_Percentage_Points"].mean(), 2),
    "percentage points"
)
```

```python
mcv_year = (
    mcv_pivot.groupby("Year", as_index=False)["MCV_Dropoff_Percentage_Points"]
    .mean()
)

plt.figure(figsize=(10, 5))
plt.plot(
    mcv_year["Year"],
    mcv_year["MCV_Dropoff_Percentage_Points"],
    marker="o"
)

plt.title("Average MCV1 to MCV2 Drop-off by Year")
plt.xlabel("Year")
plt.ylabel("Drop-off (Percentage Points)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

display(mcv_year)
```

### Insight
The MCV1 to MCV2 drop-off shows the decline in coverage between the first and second measles-containing vaccine doses. This helps identify incomplete vaccination schedules and possible follow-up gaps.

```python
# Create WHO region mapping
region_map = (
    intro[["ISO_3_Code", "WHO_Region"]]
    .drop_duplicates()
)

# Merge region information with coverage data
coverage_region = coverage.merge(
    region_map,
    left_on="Code",
    right_on="ISO_3_Code",
    how="left"
)

# Calculate average vaccination coverage by WHO region
region_coverage = (
    coverage_region.groupby("WHO_Region", as_index=False)["Coverage"]
    .mean()
    .sort_values("Coverage", ascending=False)
)

display(region_coverage)
```

```python
disease_antigen_map = {
    "MEASLES": "MCV1",
    "POLIO": "POL3",
    "DIPHTHERIA": "DTP3",
    "PERTUSSIS": "DTP3",
    "TUBERCULOSIS": "BCG",
    "HEPATITIS_B": "HEPB3"
}

correlation_results = []

for disease, antigen in disease_antigen_map.items():

    disease_df = incidence[
        incidence["Disease"] == disease
    ][["Code", "Year", "Incidence_rate"]]

    antigen_df = (
        coverage[
            coverage["Antigen"] == antigen
        ][["Code", "Year", "Coverage"]]
        .groupby(["Code", "Year"], as_index=False)["Coverage"]
        .mean()
    )

    merged_df = disease_df.merge(
        antigen_df,
        on=["Code", "Year"],
        how="inner"
    )

    correlation = merged_df["Coverage"].corr(
        merged_df["Incidence_rate"]
    )

    correlation_results.append({
        "Disease": disease,
        "Antigen": antigen,
        "Correlation": correlation
    })

correlation_table = pd.DataFrame(correlation_results)

display(correlation_table)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    correlation_table["Disease"],
    correlation_table["Correlation"]
)

plt.title("Correlation Between Vaccination Coverage and Disease Incidence")
plt.xlabel("Disease")
plt.ylabel("Correlation Coefficient")
plt.xticks(rotation=45)

plt.axhline(0)

plt.tight_layout()
plt.show()
```

### Insight
The correlation analysis compares vaccine coverage with disease incidence for related diseases and antigens. A negative correlation suggests that higher vaccination coverage is generally associated with lower disease incidence. However, correlation does not prove causation and should be interpreted as a descriptive relationship within this synthetic educational dataset.

```python
intro_summary = (
    intro.groupby(["WHO_Region", "Year"], as_index=False)
    .agg(
        Introduced=("Intro", lambda x: (x == "Yes").sum()),
        Total=("Intro", "size")
    )
)

intro_summary["Introduction_Rate"] = (
    intro_summary["Introduced"] / intro_summary["Total"] * 100
)

display(intro_summary.head(20))
```

```python
plt.figure(figsize=(11, 6))

for region in intro_summary["WHO_Region"].unique():
    region_data = intro_summary[
        intro_summary["WHO_Region"] == region
    ]

    plt.plot(
        region_data["Year"],
        region_data["Introduction_Rate"],
        marker="o",
        label=region
    )

plt.title("Vaccine Introduction Rate by WHO Region Over Time")
plt.xlabel("Year")
plt.ylabel("Introduction Rate (%)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Insight
Vaccine introduction rates differ across WHO regions and years. This analysis helps identify disparities in the timing and pace of vaccine introduction across regions.

```python
measles_intro = intro[
    intro["Description"].str.contains(
        "measles",
        case=False,
        na=False
    )
].copy()

display(measles_intro.head())
```

```python
measles_intro_year = (
    measles_intro[
        measles_intro["Intro"] == "Yes"
    ]
    .groupby(["ISO_3_Code", "Country_Name"], as_index=False)["Year"]
    .min()
    .rename(columns={"Year": "Introduction_Year"})
)

display(measles_intro_year.head())
```

```python
measles_cases = cases[
    cases["Disease"] == "MEASLES"
][["Code", "Name", "Year", "Cases"]]

measles_before_after = measles_cases.merge(
    measles_intro_year,
    left_on="Code",
    right_on="ISO_3_Code",
    how="inner"
)

measles_before_after["Period"] = np.where(
    measles_before_after["Year"] < measles_before_after["Introduction_Year"],
    "Before Introduction",
    "After Introduction"
)

display(measles_before_after.head())
```

```python
measles_comparison = (
    measles_before_after.groupby(
        "Period",
        as_index=False
    )["Cases"]
    .mean()
)

display(measles_comparison)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    measles_comparison["Period"],
    measles_comparison["Cases"]
)

plt.title("Average Measles Cases Before vs After Vaccine Introduction")
plt.xlabel("Period")
plt.ylabel("Average Reported Cases")
plt.tight_layout()
plt.show()
```

### Insight
The before-and-after comparison evaluates whether average measles cases changed following vaccine introduction. A lower average after introduction would indicate an improvement in disease control within this synthetic dataset, though this comparison alone does not establish causality.

```python
import os

# Create folder for SQL-ready CSV files
sql_import_path = "../data/sql_import"
os.makedirs(sql_import_path, exist_ok=True)

# 1. Countries table
countries = (
    intro[["ISO_3_Code", "Country_Name", "WHO_Region"]]
    .drop_duplicates()
    .rename(columns={
        "ISO_3_Code": "country_code",
        "Country_Name": "country_name",
        "WHO_Region": "who_region"
    })
)

countries.to_csv(
    f"{sql_import_path}/countries.csv",
    index=False
)

# 2. Coverage table
coverage_sql = coverage[[
    "Code",
    "Year",
    "Antigen",
    "Antigen_description",
    "Coverage_category",
    "Coverage_category_description",
    "Target_number",
    "Dodge",
    "Coverage"
]].copy()

coverage_sql.columns = [
    "country_code",
    "year",
    "antigen",
    "antigen_description",
    "coverage_category",
    "coverage_category_description",
    "target_number",
    "doses_administered",
    "coverage"
]

coverage_sql.to_csv(
    f"{sql_import_path}/coverage_data.csv",
    index=False
)

# 3. Incidence table
incidence_sql = incidence[[
    "Code",
    "Year",
    "Disease",
    "Disease_description",
    "Denominator",
    "Incidence_rate"
]].copy()

incidence_sql.columns = [
    "country_code",
    "year",
    "disease",
    "disease_description",
    "denominator",
    "incidence_rate"
]

incidence_sql.to_csv(
    f"{sql_import_path}/incidence_rate.csv",
    index=False
)

# 4. Reported cases table
cases_sql = cases[[
    "Code",
    "Year",
    "Disease",
    "Disease_description",
    "Cases"
]].copy()

cases_sql.columns = [
    "country_code",
    "year",
    "disease",
    "disease_description",
    "cases"
]

cases_sql.to_csv(
    f"{sql_import_path}/reported_cases.csv",
    index=False
)

# 5. Vaccine introduction table
intro_sql = intro[[
    "ISO_3_Code",
    "Year",
    "Description",
    "Intro"
]].copy()

intro_sql.columns = [
    "country_code",
    "year",
    "description",
    "intro"
]

intro_sql.to_csv(
    f"{sql_import_path}/vaccine_introduction.csv",
    index=False
)

# 6. Vaccine schedule table
schedule_sql = schedule[[
    "ISO_3_Code",
    "Year",
    "Vaccine_code",
    "Vaccine_description",
    "Schedule_rounds",
    "Target_pop",
    "Target_pop_description",
    "Geoarea",
    "Age_administered",
    "Source_comment"
]].copy()

schedule_sql.columns = [
    "country_code",
    "year",
    "vaccine_code",
    "vaccine_description",
    "schedule_rounds",
    "target_pop",
    "target_pop_description",
    "geoarea",
    "age_administered",
    "source_comment"
]

schedule_sql.to_csv(
    f"{sql_import_path}/vaccine_schedule.csv",
    index=False
)

print("SQL import files created successfully.")
```
